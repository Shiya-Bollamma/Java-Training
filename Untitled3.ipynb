```python
import torch
from transformers import pipeline

# Load the model directly to the Colab GPU
print("Loading model onto GPU...")
generator = pipeline(
    "text-generation",
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    torch_dtype=torch.float16,
    device_map="auto"
)

def run_agent(role: str, desire: str, content: str) -> str:
    prompt = f"""<|system|>
You are a {role}. Your primary goal is to {desire}.</s>
<|user|>
Context/Input data: {content}
Please process this according to your role and goal.</s>
<|assistant|>"""

    outputs = generator(
        prompt,
        max_new_tokens=256,
        temperature=0.7,
        do_sample=True,
        top_k=50,
        top_p=0.95
    )

    full_text = outputs[0]['generated_text']
    return full_text.split("<|assistant|>")[-1].strip()
```

```python
# @title Interactive Agent Inputs { display-mode: "form" }

Role = "VTU AI/ML Professor"  # @param {type:"string"}
Desire = "Explain the topic with a real-world example and key points"  # @param {type:"string"}
Content = "Packet switching splits data into smaller units called packets that travel independently across a network."  # @param {type:"string"}

# Execute text generation
output = run_agent(Role, Desire, Content)
print("\n=== AI OUTPUT ===")
print(output)
```

```python
!pip install -q gradio

import gradio as gr

def process_agent(role, desire, content):
    return run_agent(role, desire, content)

# Create a web interface
demo = gr.Interface(
    fn=process_agent,
    inputs=[
        gr.Textbox(lines=2, placeholder="e.g., Senior Software Engineer", label="Role"),
        gr.Textbox(lines=2, placeholder="e.g., Explain the bug and suggest a fix", label="Desire"),
        gr.Textbox(lines=5, placeholder="Paste your input text or code here...", label="Content")
    ],
    outputs=gr.Textbox(label="Model Response"),
    title="Role-Based Transformer Agent",
    description="Dynamically alters language model persona and objective based on structured input variables."
)

demo.launch(share=True)  # Generates a public URL to share and test the live app
```
